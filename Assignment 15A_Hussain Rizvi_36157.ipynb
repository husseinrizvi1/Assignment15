```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\36157\Downloads\ride_hailing_dataset.csv")
```

```python
print(df.head())
```

```python
avg_fare = df.groupby('service_type')['fare'].mean().sort_values(ascending=False)
print(avg_fare.round(2))
```

```python
city_counts = df['city'].value_counts()
print(city_counts)
```

```python
## Islamabad has the highest number of trips with 29 out of 100
```

```python
avg_distance = df['distance_km'].mean()
print(f"Average Distance per Trip: {avg_distance:.2f} km")
```

```python
rating = df.groupby('gender')['trip_rating'].agg(['count', 'mean', 'median', 'std'])
print(rating.round(2))
```

```python
plt.figure(figsize=(8, 5))
plt.bar(avg_fare.index, avg_fare.values)
plt.title('Average Fare by Service Type')
plt.xlabel('Service Type')
plt.ylabel('Average Fare (PKR)')
plt.show()
```

```python
## The bar chart illustrates the average fare across the four ride-hailing service types. InDrive has the highest average fare (PKR 573.51), followed by Careem (PKR 544.98). Rickshaw (PKR 486.64) and Uber (PKR 486.13) have nearly identical, lower average fares—roughly PKR 87 less per trip than InDrive.
```
